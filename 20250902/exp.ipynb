```python
import pyvisa
```

```python
rm = pyvisa.ResourceManager('C:/windows/System32/visa64.dll')
print(rm.list_resources())
```

```python
smu = rm.open_resource('TCPIP0::169.254.0.1::inst0::INSTR')
```

```python
smu.timeout = 20000
```

```python
file_path = "C:\\Users\\20245580\\work\\Automate_Lab_Instrument\\20250902\\"
file_name = "ex_1.tsp" 
file_path_and_name = file_path + file_name 
print(file_path_and_name)

```

```python
smu.write("loadscript Example1")
```

```python

with open(file_path_and_name) as fp:
    for line in fp: smu.write(line)
```

```python
smu.write("endscript") 
```

```python
smu.write("Example1.run()") 
```

```python
smu.close() # through LAN cable, must close it, before open another connection
```
