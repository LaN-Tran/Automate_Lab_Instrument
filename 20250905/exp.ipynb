# upload tsp file to SMU and run

Import the `ex_1.tsp` script into the keithley internal memory and run it

- [1]  upload `.tsp` script to Keithley SMU through python visa: https://www.tek.com/de/documents/application-note/harness-the-power-of-tsp-toolkit-software - see **Application Example**

- [2]  managing scripts loaded in the Keithley SMU: **2600BS-901-01F_2600B_Reference_Aug2021.pdf- section: Create and load a script**

- [3] https://github.com/tektronix/keithley/blob/main/Instrument_Examples/Series_2600/2602/ASimpleTest/main.tsp

## M1

```python
import pyvisa
```

```python
rm = pyvisa.ResourceManager('C:/windows/System32/visa64.dll')
print(rm.list_resources())
```

```python
smu = rm.open_resource('USB0::0x05E6::0x2636::4480001::INSTR')
```

```python
smu.timeout = 20000
```

### upload example 1 tsp code

```python
file_path = "C:\\Users\\20245580\\LabCode\\Automate_Lab_Instrument\\20250905\\"
file_name = "pulse_trains_2chs.tsp" 
file_path_and_name = file_path + file_name 
print(file_path_and_name)

```

```python
smu.write("loadscript Example1")
```

```python

with open(file_path_and_name) as fp:
    for line in fp: smu.write(line)
```

```python
smu.write("endscript") 
```

```python
smu.write("Example1.run()") 
```

```python
smu.close() # through LAN cable, must close it, before open another connection
```
